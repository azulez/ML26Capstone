# Example Notebook

A quick test notebook to check rendering in VSCode.

```python
import numpy as np

x = np.arange(10)
y = x ** 2
print(x)
print(y)
```

## Plot

```python
import matplotlib.pyplot as plt

plt.plot(x, y, marker="o")
plt.title("y = x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
